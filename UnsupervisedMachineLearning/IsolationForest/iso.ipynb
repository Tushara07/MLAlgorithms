```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest

import pickle

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("creditcard.csv")
```

```python
X = df.drop(
    "Class",
    axis=1
)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
model = IsolationForest(
    contamination=0.002,
    random_state=42
)
```

```python
model.fit(
    X_scaled
)
```

```python
predictions = model.predict(
    X_scaled
)
```

```python
predictions = np.where(
    predictions == -1,
    1,
    0
)
```

```python
df["Predicted_Class"] = predictions
```

```python
df["Predicted_Class"].value_counts()
```

```python
pd.crosstab(
    df["Class"],
    df["Predicted_Class"]
)
```

```python
actual_frauds = (
    df["Class"] == 1
).sum()

predicted_frauds = (
    df["Predicted_Class"] == 1
).sum()

print(
    "Actual Frauds:",
    actual_frauds
)

print(
    "Predicted Frauds:",
    predicted_frauds
)
```

```python
plt.figure(
    figsize=(5,3)
)

df["Predicted_Class"].value_counts().plot(
    kind="bar"
)

plt.xlabel(
    "Class"
)

plt.ylabel(
    "Count"
)

plt.title(
    "Isolation Forest Predictions"
)

plt.show()
```

```python
plt.figure(
    figsize=(6,4)
)

plt.hist(
    df["Amount"],
    bins=50
)

plt.xlabel(
    "Transaction Amount"
)

plt.ylabel(
    "Frequency"
)

plt.title(
    "Transaction Amount Distribution"
)

plt.show()
```

```python
df.groupby(
    "Predicted_Class"
)["Amount"].mean()
```

```python
fraud_names = {
    0: "Normal Transaction",
    1: "Potential Fraud"
}
```

```python
df["Transaction_Type"] = (
    df["Predicted_Class"]
    .map(fraud_names)
)
```

```python
df.to_csv(
    "fraud_detection_results.csv",
    index=False
)
```
